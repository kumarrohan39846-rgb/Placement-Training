Numpy Array Operation

```python
import numpy as np
```

```python
import numpy as np
arr=np.array([[10, 20, 30, 40, 50]])
print("Basic Slicing:",arr[:4])
print("Slicing with step:",arr[0:5:2])
print("Negative Indexing:",arr[-1])
```

```python
arr_2d=np.array([[1, 2, 3, 10], [4, 5, 6, 20],[7, 8, 9, 30]])
print("Specific element:",arr_2d[1, 2])
print("Entire row:\n",arr_2d[1,:])
print("Entire column:\n",arr_2d[:,1:])
print("Last 2 years sales:",arr_2d[2,2:])
```

Sorting

```python
arr_1d=np.array([3,5,2,9,5,6,1,7])
print("Sorting 1d Array",sorted(arr_1d))
print("Sorting 1d Array",np.sort(arr_1d))
arr_2d=np.array([[2,1],[3,2],[4,3]])
# Axis=0->col axis=1->row
print("Sorting 2d Array:\n",np.sort(arr_2d,axis=0))
```

#Filter

```python
number=np.array([3,5,2,9,5,6,1,7])
even_number=number[number%2==0]
print("Even numbers:",even_number)
```

##Filter with mask

```python
marks=np.array([67,34,10,23,50,97,54])
mask_pass=marks>35
print("More than 35(mask):",mask_pass)
```
