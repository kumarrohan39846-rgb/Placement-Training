dict functions and method

```python
student={
    'id':"S2601",
    'name':"John",
    'age':20,
    'dept':"CSEB"
}
#retrieve a single value using key
get_values=student.get('age')
print(get_values)
#check all the key
dict_keys=student.keys()
print(dict_keys)
#retrieve all the values
dict_values=student.values()
print(dict_values)
#update
update_age=student.update({'age':21})
print(student)
#setdefault
dep_val=student.setdefault('dept','CS')
print(dep_val)
new_loc=student.setdefault("place","Bangalore")
print(student)
#copy
c=student.copy()
print(c)
#pop remove a key value pair
student.pop("dept")
print(student)
#popitem
student.popitem()
print(student)
#dice_items
print(student.items())
```

dict iteration

```python
student={
    'id':"S2601",
    'name':"John",
    'age':20,
    'dept':"CSEB"
}
#itr over keys
for k in student:
    print(k)
#itr over values
for v in student.values():
    print(v)
#itr over key-value pairs
for k, v in student.items():
    print(k,":" ,v)
```

Write a script To generate a lookup dictionary where keys represent risk tiers(1 to 15) and values are the square of each tier number

```python
dict_square={}
for i in range(1,16):
    dict_square.setdefault(i,i**2)
print(dict_square)
```
